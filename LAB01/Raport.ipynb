1. Çfarë është Markdown?

Markdown është një gjuhë e thjeshtë formatimi që përdoret në Jupyter Notebook për të shkruar dhe organizuar tekstin.

2. Përdorimi i Markdown

Markdown na mundëson të krijojmë:

Tituj dhe nëntituj.
Tekst të trashë dhe të pjerrët.
Lista me pika dhe numra.
Lidhje dhe imazhe.
3. Shembuj të Markdown

Krijimi i titujve:

markdown
# Titulli kryesor
## Nëntitulli
### Titulli i vogël


Teksti i trashë dhe i pjerrët:

markdown
**Tekst i trashë**

*Tekst i pjerrët*


Krijimi i listave:

markdown
- Python
- Java
- C++


Lista me numra:

markdown
1. Hapi i parë
2. Hapi i dytë
3. Hapi i tretë

4. Përfundim

Markdown është shumë i dobishëm në Jupyter Notebook, sepse na ndihmon të organizojmë dhe të formatojmë tekstin në mënyrë të thjeshtë dhe të qartë.

5. Si përdoret Markdown në Jupyter Notebook?

Për të përdorur Markdown në Jupyter Notebook duhet të ndjekim këto hapa:

Hapim Jupyter Notebook.
Krijojmë një qelizë të re (Cell).
Zgjedhim opsionin Markdown.
Shkruajmë tekstin ose komandat e Markdown.
Shtypim Shift + Enter për të shfaqur rezultatin.
6. Krijimi i vijave ndarëse

Markdown mund të përdoret edhe për të krijuar vija ndarëse midis pjesëve të tekstit.

7. Shtimi i lidhjeve

Me Markdown mund të shtojmë lidhje në faqe të ndryshme interneti.

Shembull:

markdown
[Google](https://www.google.com)

8. Shtimi i tabelave

Markdown na lejon të krijojmë edhe tabela për organizimin e informacionit.

Shembull:

markdown
| Emri | Lënda |
|------|-------|
| Kristi | Informatikë |
| Ana | Matematikë |

9. Përparësitë e Markdown

Disa nga përparësitë kryesore janë:

Është i thjeshtë për t'u mësuar.
Kursen kohë gjatë shkrimit.
E bën dokumentin më të organizuar.
Ndihmon në paraqitjen e informacionit.
Përdoret lehtësisht në Jupyter Notebook.
10. Përfundimi

Markdown është një mjet shumë i rëndësishëm në Jupyter Notebook. Ai na ndihmon të shkruajmë, formatojmë dhe organizojmë informacionin në mënyrë të thjeshtë dhe të kuptueshme. Duke përdorur Markdown, dokumentet tona bëhen më të bukura dhe më profesionale.
